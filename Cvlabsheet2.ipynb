```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
filename = list(uploaded.keys())[0]

img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

print("Image shape:", img.shape)
print("Image loaded successfully!")
```

```python
plt.figure(figsize=(8, 6))

plt.imshow(img, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(img.ravel(), bins=256, range=[0, 256])

plt.title("Histogram of Original Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.show()
```

```python
# Contrast Stretching
min_val = np.min(img)
max_val = np.max(img)

contrast_stretched = ((img - min_val) / (max_val - min_val) * 255).astype(np.uint8)

plt.figure(figsize=(8, 6))

plt.imshow(contrast_stretched, cmap='gray')
plt.title("Contrast Stretched Image")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(contrast_stretched.ravel(), bins=256, range=[0, 256])

plt.title("Histogram of Contrast Stretched Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.show()
```

```python
# Histogram Equalization
equalized = cv2.equalizeHist(img)

plt.figure(figsize=(8, 6))

plt.imshow(equalized, cmap='gray')
plt.title("Histogram Equalized Image")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(equalized.ravel(), bins=256, range=[0, 256])

plt.title("Histogram of Equalized Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.show()
```

```python
# CLAHE
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

clahe_img = clahe.apply(img)

plt.figure(figsize=(8, 6))

plt.imshow(clahe_img, cmap='gray')
plt.title("CLAHE Enhanced Image")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(clahe_img.ravel(), bins=256, range=[0, 256])

plt.title("Histogram of CLAHE Enhanced Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(16, 10))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(contrast_stretched, cmap='gray')
plt.title("Contrast Stretching")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(equalized, cmap='gray')
plt.title("Histogram Equalization")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(clahe_img, cmap='gray')
plt.title("CLAHE")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 10))

plt.subplot(2, 2, 1)
plt.hist(img.ravel(), bins=256, range=[0, 256])
plt.title("Original Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 2)
plt.hist(contrast_stretched.ravel(), bins=256, range=[0, 256])
plt.title("Contrast Stretching Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 3)
plt.hist(equalized.ravel(), bins=256, range=[0, 256])
plt.title("Histogram Equalization")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 4)
plt.hist(clahe_img.ravel(), bins=256, range=[0, 256])
plt.title("CLAHE Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
print("Original Image")
print("Minimum intensity:", np.min(img))
print("Maximum intensity:", np.max(img))
print("Mean intensity:", np.mean(img))

print("\nContrast Stretched Image")
print("Minimum intensity:", np.min(contrast_stretched))
print("Maximum intensity:", np.max(contrast_stretched))
print("Mean intensity:", np.mean(contrast_stretched))

print("\nHistogram Equalized Image")
print("Minimum intensity:", np.min(equalized))
print("Maximum intensity:", np.max(equalized))
print("Mean intensity:", np.mean(equalized))

print("\nCLAHE Image")
print("Minimum intensity:", np.min(clahe_img))
print("Maximum intensity:", np.max(clahe_img))
print("Mean intensity:", np.mean(clahe_img))
```

**Q&A**

**Question-1** What is image contrast, and why is it important in image processing?

**Answer-**
Image contrast is the difference in intensity or brightness between different regions of an image. An image with high contrast has clearly distinguishable bright and dark areas, while a low-contrast image has pixel intensities concentrated within a narrow range.

Contrast is important because it improves the visibility of objects, edges, and details. Proper contrast enhancement can make important features easier to identify and can improve the performance of subsequent computer vision tasks such as segmentation, feature extraction, and object recognition.

**Question-2** Explain the concept of an image histogram. What information does it provide?

**Answer-** An image histogram is a graphical representation of the distribution of pixel intensity values in an image.

For a grayscale image, the horizontal axis represents intensity values, usually from 0 to 255, while the vertical axis represents the number of pixels having each intensity.

It provides information about:

* Brightness distribution
* Contrast of the image
* Dark and bright regions
* Whether pixel values are concentrated in a narrow range
* Overall intensity distribution

A narrow histogram usually indicates low contrast, while a histogram spread across a wider intensity range generally indicates better contrast.


**Question-4** What is CLAHE? How does it differ from standard Histogram Equalization?

**Answer-** CLAHE stands for Contrast Limited Adaptive Histogram Equalization.

It is an adaptive contrast enhancement technique that divides an image into small regions called tiles and performs histogram equalization on each region separately. It also limits excessive contrast enhancement using a clip limit.

The major difference is:

* Standard Histogram Equalization: Works on the entire image globally.
* CLAHE: Works on small local regions and limits excessive enhancement.

Therefore, CLAHE can provide better results when different areas of an image have different illumination levels. The practical specifically asks you to observe CLAHE on images with varying illumination.


**Questiion-5** Why is histogram equalization commonly applied before feature extraction and image segmentation?

**Answer-** Histogram equalization can improve the contrast of an image and make important features such as edges, textures, and object boundaries more visible.


This can help feature extraction algorithms identify useful characteristics more effectively. Similarly, improved contrast can make it easier for segmentation methods to distinguish objects from their backgrounds.


Therefore, histogram equalization can serve as a preprocessing step before feature extraction and segmentation.

**Answer-**  Three applications are:
* Medical imaging – enhancing X-rays, CT scans, and other medical images.
* Satellite and remote-sensing images – improving visibility of geographical features.
* Low-light photography – enhancing details in dark images.

These types of images are also specifically mentioned as suitable real-world examples in the practical sheet.


**Question-7** What are the limitations of global histogram equalization?

**Answer-** The limitations include:

* It applies the same enhancement to the entire image.
* It may over-enhance noise.
* It can produce unnatural-looking results.
* It may lose some local details.
* It does not handle regions with different illumination particularly well.
* Very bright or dark regions can sometimes become excessively enhanced.

his is one reason adaptive methods such as CLAHE can be useful.

**Question-8** How does contrast enhancement improve the performance of object detection and recognition systems?

**Answer-** Contrast enhancement improves the distinction between objects and their surroundings. This can make important features such as edges, shapes, textures, and boundaries more visible.

As a result, image-processing and computer-vision algorithms may have an easier time extracting useful features and identifying objects.

The experiment notes that poor contrast can make feature extraction and object recognition difficult, while histogram-based enhancement improves pixel-intensity distribution and image visibility.


**Question-9** Why is CLAHE preferred for medical imaging and low-light image enhancement?

**Answer-** **CLAHE** is useful for medical and low-light images because these images can contain different illumination levels in different regions.

Instead of treating the entire image uniformly, CLAHE enhances contrast locally while limiting excessive enhancement. This helps reveal details in dark or poorly illuminated areas while reducing the possibility of extreme contrast enhancement.
